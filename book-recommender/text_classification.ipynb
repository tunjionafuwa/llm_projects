```python
category_mapping = {
    "Fiction": "Fiction",
    "Juvenile Fiction": "Children's Fiction",
    "Biography & Autobiography": "Nonfiction",
    "History": "Nonfiction",
    "Literary Criticism	": "Nonfiction",
    "Philosophy": "Nonfiction",
    "Religion": "Nonfiction",
    "Comics & Graphic Novels": "Fiction",
    "Drama": "Fiction",
    "Juvenile Nonfiction": "Children's Nonfiction",
    "Science": "Nonfiction",
    "Poetry": "Fiction",
}

books["simple_categories"] = books["categories"].map(category_mapping)
```

```python
books[books["simple_categories"].isna()]
```

```python
from transformers import pipeline

fiction_categories = ["Fiction", "Nonfiction"]
classifier = pipeline(
    "zero-shot-classification", model="facebook/bart-large-mnli", device="mps"
)
```

```python
classifier(
    books.loc[books["simple_categories"] == "Fiction", "description"].reset_index(
        drop=True
    )[0],
    fiction_categories,
)
```

```python
import numpy as np


def generate_predictions(sequence: str, categories: list[str] = fiction_categories):
    pred = classifier(sequences=sequence, candidate_labels=categories)
    max_index = np.argmax(pred["scores"])
    return pred["labels"][max_index]
```

```python
from tqdm import tqdm

pred_cats = []
actual_cats = []

for i in tqdm(range(300)):
    sequence = books.loc[
        books["simple_categories"] == "Fiction", "description"
    ].reset_index(drop=True)[i]
    pred_cats += [generate_predictions(sequence, fiction_categories)]
    actual_cats += ["Fiction"]

for i in tqdm(range(300)):
    sequence = books.loc[
        books["simple_categories"] == "Nonfiction", "description"
    ].reset_index(drop=True)[i]
    pred_cats += [generate_predictions(sequence, fiction_categories)]
    actual_cats += ["Nonfiction"]
```

```python
pred_df = pd.DataFrame({"actual_cat": actual_cats, "pred_cat": pred_cats})
pred_df["correct_pred"] = np.where(pred_df["actual_cat"] == pred_df["pred_cat"], 1, 0)
```

```python
((np.sum(pred_df["correct_pred"].values) / len(pred_df)) * 100).item()
```

```python
missing_cat = books.loc[
    books["simple_categories"].isna(), ["isbn13", "description"]
].reset_index(drop=True)
```

```python
pred_cats = []
isbn13 = []

for i in tqdm(range(len(missing_cat))):
    sequence = missing_cat["description"][i]
    pred_cats += [generate_predictions(sequence, fiction_categories)]
    isbn13 += [missing_cat["isbn13"][i]]
```

```python
missing_cat_df = pd.DataFrame({"isbn13": isbn13, "pred_cat": pred_cats})
books = pd.merge(books, missing_cat_df, on="isbn13", how="left")
```

```python
books["simple_categories"] = np.where(
    books["simple_categories"].isna(), books["pred_cat"], books["simple_categories"]
)
books = books.drop(columns=["pred_cat"])
```
