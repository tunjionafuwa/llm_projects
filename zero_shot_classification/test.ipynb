```python
from transformers import pipeline
```

```python

classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli", device="mps")
```

```python
sequence = "The company's quarterly earnings increased by 20%, exceeding market expectations."
candidate_labels = ["finance", "sports", "politics", "technology"]
cc = classifier(sequence, candidate_labels)
cc
```
